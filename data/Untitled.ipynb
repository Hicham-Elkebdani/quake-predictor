```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

def convolution_2d(img: np.ndarray, kernel: np.ndarray) -> np.ndarray:
    h_img, w_img = img.shape
    h_k, w_k = kernel.shape

    kx = h_k // 2
    ky = w_k // 2

    out_h = h_img - 2 * kx
    out_w = w_img - 2 * ky
    output = np.zeros((out_h, out_w), dtype=float)

    for i in range(kx, h_img - kx):
        for j in range(ky, w_img - ky):

            region = img[i - kx : i + kx + 1, j - ky : j + ky + 1]

            value = np.sum(region * kernel)

            output[i - kx, j - ky] = value

    output = np.clip(output, 0, 255)
    return output.astype(np.uint8)



img = cv2.imread("C:\\Users\\pc\\.dbus-keyrings\\Desktop\\Vision\\TP2\\d.jpg", cv2.IMREAD_GRAYSCALE)

kernel = np.array([[1,1,1],[1,1,1],[1,1,1]], dtype=float)/9

output = convolution_2d(img, kernel)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.subplot(1,2,2)
plt.imshow(output, cmap='gray')
plt.axis('off')
plt.show()
```
